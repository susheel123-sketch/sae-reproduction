```python
from datasets import load_dataset

ds = load_dataset("NeelNanda/pile-10k", split="train")
print(len(ds))
```

```python
all_acts = []
target_vectors = 500_000
stride = 8

for i, row in enumerate(ds):
    if sum(a.shape[0] for a in all_acts) >= target_vectors:
        break
    tokens = model.to_tokens(row["text"])[:, :512]
    with torch.no_grad():
        _, cache = model.run_with_cache(tokens, names_filter="blocks.6.hook_resid_post")
    acts = cache["blocks.6.hook_resid_post"][0, ::stride, :].cpu()
    all_acts.append(acts)
    if i % 200 == 0:
        print(f"doc {i}, total vectors so far: {sum(a.shape[0] for a in all_acts)}")

activations = torch.cat(all_acts, dim=0)
print("FINAL SHAPE:", activations.shape)
torch.save(activations, "data/activations.pt")
```

```python
import torch.nn as nn

class SAE(nn.Module):
    def __init__(self, d_in=768, d_hidden=6144):
        super().__init__()
        self.encoder = nn.Linear(d_in, d_hidden)
        self.decoder = nn.Linear(d_hidden, d_in, bias=False)
        self.b_dec = nn.Parameter(torch.zeros(d_in))

    def forward(self, x):
        x_centered = x - self.b_dec
        features = torch.relu(self.encoder(x_centered))
        recon = self.decoder(features) + self.b_dec
        return recon, features
```

```python
def loss_fn(x, recon, features, l1_coef=1e-3):
    mse = ((x - recon) ** 2).mean()
    l1 = features.abs().sum(dim=-1).mean()
    return mse + l1_coef * l1, mse, l1

sae = SAE().to(device)
optimizer = torch.optim.Adam(sae.parameters(), lr=1e-3)

data = torch.load("data/activations.pt").to(device)
tiny = data[:2000]

for epoch in range(3):
    recon, features = sae(tiny)
    loss, mse, l1 = loss_fn(tiny, recon, features)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"epoch {epoch}: loss={loss.item():.4f}, mse={mse.item():.4f}, l1={l1.item():.4f}")
```

```python
# Normalize: scale activations so their average L2 norm equals sqrt(d_model)
import math

d_model = data.shape[1]
avg_norm = data.norm(dim=-1).mean()
scale_factor = math.sqrt(d_model) / avg_norm
data_normalized = data * scale_factor

print(f"avg norm before: {avg_norm:.2f}, scale factor: {scale_factor:.4f}")
print(f"avg norm after: {data_normalized.norm(dim=-1).mean():.2f}")
```

```python
sae = SAE().to(device)
optimizer = torch.optim.Adam(sae.parameters(), lr=3e-4)  # lowered from 1e-3

tiny = data_normalized[:2000]

for epoch in range(5):
    recon, features = sae(tiny)
    loss, mse, l1 = loss_fn(tiny, recon, features)
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    print(f"epoch {epoch}: loss={loss.item():.4f}, mse={mse.item():.4f}, l1={l1.item():.4f}")
```

```python
sae = SAE().to(device)
optimizer = torch.optim.Adam(sae.parameters(), lr=3e-4)
batch_size = 1024

for epoch in range(5):
    perm = torch.randperm(data_normalized.shape[0])
    total_loss = 0
    total_mse = 0
    total_l1 = 0
    n_batches = 0
    for i in range(0, data_normalized.shape[0], batch_size):
        batch = data_normalized[perm[i:i+batch_size]]
        recon, features = sae(batch)
        loss, mse, l1 = loss_fn(batch, recon, features)
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        total_mse += mse.item()
        total_l1 += l1.item()
        n_batches += 1
    print(f"epoch {epoch}: avg_loss={total_loss/n_batches:.4f}, avg_mse={total_mse/n_batches:.4f}, avg_l1={total_l1/n_batches:.4f}")

torch.save(sae.state_dict(), "checkpoints/sae.pt")
```

```python
sae.eval()
with torch.no_grad():
    recon, features = sae(data_normalized)
    mse = ((data_normalized - recon) ** 2).mean().item()

    # L0 sparsity: average number of active (nonzero) features per token
    l0 = (features > 0).float().sum(dim=-1).mean().item()

    # Variance explained: how much of the original signal is captured
    total_var = data_normalized.var(dim=0).sum().item()
    residual_var = (data_normalized - recon).var(dim=0).sum().item()
    variance_explained = 1 - (residual_var / total_var)

print(f"Final MSE: {mse:.4f}")
print(f"Average L0 (active features per token): {l0:.1f}")
print(f"Variance explained: {variance_explained:.4f}")
print(f"Dictionary size: {features.shape[1]}")
```

```python
from datasets import load_dataset

ds = load_dataset("NeelNanda/pile-10k", split="train")

all_acts = []
target_vectors = 500_000
stride = 8

for i, row in enumerate(ds):
    if sum(a.shape[0] for a in all_acts) >= target_vectors:
        break
    tokens = model.to_tokens(row["text"])[:, :512]
    with torch.no_grad():
        _, cache = model.run_with_cache(tokens, names_filter="blocks.6.hook_resid_post")
    acts = cache["blocks.6.hook_resid_post"][0, ::stride, :].cpu()
    all_acts.append(acts)

activations = torch.cat(all_acts, dim=0)
print("FINAL SHAPE:", activations.shape)
torch.save(activations, "data/activations.pt")
```

```python
def get_loss_recovered(model, sae, text, hook_name="blocks.6.hook_resid_post", scale_factor=1.0):
    tokens = model.to_tokens(text)

    clean_loss = model(tokens, return_type="loss").item()

    def zero_hook(act, hook):
        return torch.zeros_like(act)
    zero_loss = model.run_with_hooks(tokens, return_type="loss",
                                       fwd_hooks=[(hook_name, zero_hook)]).item()

    def sae_hook(act, hook):
        act_normalized = act * scale_factor
        recon, _ = sae(act_normalized)
        return recon / scale_factor
    recon_loss = model.run_with_hooks(tokens, return_type="loss",
                                        fwd_hooks=[(hook_name, sae_hook)]).item()

    loss_recovered = 1 - (recon_loss - clean_loss) / (zero_loss - clean_loss)
    return clean_loss, zero_loss, recon_loss, loss_recovered

test_texts = [
    "The history of artificial intelligence dates back to ancient myths.",
    "Climate change is one of the most pressing issues of our time.",
    "The stock market experienced significant volatility this week.",
]

sae.eval()
for text in test_texts:
    clean, zero, recon, recovered = get_loss_recovered(model, sae, text, scale_factor=scale_factor.item())
    print(f"clean={clean:.3f}, zero_ablation={zero:.3f}, recon={recon:.3f}, loss_recovered={recovered*100:.1f}%")
```
